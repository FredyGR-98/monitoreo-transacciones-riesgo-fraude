# Analisis de riesgo transaccional basado en reglas de negocio

Este notebook desarrolla un analisis de riesgo transaccional para fraude bancario utilizando el dataset PaySim.

El objetivo no es entrenar modelos predictivos, sino crear un pipeline simple y explicable que permita:

- generar variables derivadas relevantes,
- aplicar reglas de negocio,
- calcular un puntaje de riesgo por transaccion,
- clasificar el nivel de riesgo,
- producir una tabla final lista para analisis, Power BI o dashboard.

La filosofia del analisis es KISS: reglas simples, transparentes y faciles de justificar desde negocio.

## 2. Creacion de variables derivadas

Las variables derivadas transforman los campos originales en senales interpretables de riesgo.

En esta etapa se priorizan variables faciles de explicar:

- relacion entre monto y saldo disponible,
- vaciamiento de cuenta de origen,
- consistencia de balances antes y despues de la transaccion,
- transacciones de monto alto en transferencias,
- identificacion explicita de operaciones tipo transferencia.

```python
def crear_variables_derivadas(datos: pd.DataFrame) -> tuple[pd.DataFrame, dict]:
    """Crea variables explicables para analisis de riesgo.

    La funcion no elimina registros ni modifica el dataset original. Devuelve
    una copia enriquecida y un diccionario con umbrales usados, para que la
    logica sea auditable.
    """
    df = datos.copy()

    # Regla de negocio: una transferencia de monto alto se define usando el
    # percentil 95 de los montos en operaciones TRANSFER. Este umbral es simple,
    # adaptable al dataset y facil de explicar.
    montos_transfer = df.loc[df["type"].eq("TRANSFER"), "amount"]
    umbral_monto_alto_transfer = (
        montos_transfer.quantile(0.95)
        if not montos_transfer.empty
        else df["amount"].quantile(0.95)
    )

    # Evitamos divisiones por cero. Cuando no hay saldo de origen positivo,
    # el ratio queda como NaN porque no existe una base valida de comparacion.
    df["ratio_monto_saldo"] = np.where(
        df["oldbalanceOrg"] > 0,
        df["amount"] / df["oldbalanceOrg"],
        np.nan,
    )

    # Senal de vaciamiento: la cuenta queda practicamente en cero despues de la
    # operacion. Se usa tolerancia pequena para evitar problemas de decimales.
    df["flag_vaciamiento"] = (
        (df["oldbalanceOrg"] > 0)
        & np.isclose(df["newbalanceOrig"], 0, atol=0.01)
    ).astype(int)

    # Diferencias de balance: comparan lo que deberia ocurrir contablemente con
    # lo registrado despues de la transaccion.
    df["diferencia_balance_origen"] = (
        (df["oldbalanceOrg"] - df["amount"]) - df["newbalanceOrig"]
    ).abs()
    df["diferencia_balance_destino"] = (
        (df["oldbalanceDest"] + df["amount"]) - df["newbalanceDest"]
    ).abs()

    # Flags simples que luego seran usados por las reglas de riesgo.
    df["flag_transfer"] = df["type"].eq("TRANSFER").astype(int)
    df["flag_monto_alto"] = (
        df["type"].eq("TRANSFER")
        & (df["amount"] >= umbral_monto_alto_transfer)
    ).astype(int)

    umbrales = {
        "percentil_monto_alto_transfer": 0.95,
        "umbral_monto_alto_transfer": umbral_monto_alto_transfer,
        "tolerancia_balance": 1.0,
        "umbral_ratio_alto": 0.90,
    }

    return df, umbrales


datos_riesgo, umbrales = crear_variables_derivadas(datos)
umbrales
```

**Interpretacion:** las variables derivadas convierten la informacion transaccional en senales de riesgo entendibles para negocio.

Por ejemplo, un ratio alto entre monto y saldo indica que la transaccion consume una parte relevante del saldo disponible, mientras que el vaciamiento de cuenta puede representar un comportamiento sospechoso si ocurre junto con una transferencia de alto monto.

```python
# Vista rapida de variables derivadas principales.
columnas_derivadas = [
    "type",
    "amount",
    "oldbalanceOrg",
    "newbalanceOrig",
    "ratio_monto_saldo",
    "flag_vaciamiento",
    "diferencia_balance_origen",
    "diferencia_balance_destino",
    "flag_transfer",
    "flag_monto_alto",
    "isFraud",
]

datos_riesgo[columnas_derivadas].head()
```

## 3. Definicion de reglas de riesgo

Las reglas de riesgo son condiciones binarias que se activan cuando una transaccion cumple un patron sospechoso.

Cada regla esta pensada para ser interpretable:

- monto alto en transferencia,
- ratio alto entre monto y saldo,
- vaciamiento de cuenta,
- inconsistencias de balance,
- combinacion de multiples senales.

```python
def aplicar_reglas_riesgo(datos: pd.DataFrame, umbrales: dict) -> pd.DataFrame:
    """Aplica reglas simples e interpretables de riesgo.

    Cada regla genera una columna binaria donde 1 significa que la senal se
    activo para la transaccion y 0 significa que no se activo.
    """
    df = datos.copy()

    # Regla 1: transferencia de monto alto segun percentil 95 de TRANSFER.
    df["regla_monto_alto_transfer"] = df["flag_monto_alto"]

    # Regla 2: el monto representa al menos el 90% del saldo de origen.
    # Esta senal captura transacciones que consumen casi todo el saldo disponible.
    df["regla_ratio_alto"] = (
        (df["oldbalanceOrg"] > 0)
        & (df["ratio_monto_saldo"] >= umbrales["umbral_ratio_alto"])
    ).astype(int)

    # Regla 3: vaciamiento de cuenta de origen.
    df["regla_vaciamiento"] = df["flag_vaciamiento"]

    # Regla 4: inconsistencias contables en balances.
    # Se usa tolerancia para evitar marcar diferencias minimas por redondeo.
    tolerancia = umbrales["tolerancia_balance"]
    df["regla_inconsistencia_origen"] = (
        df["diferencia_balance_origen"] > tolerancia
    ).astype(int)

    # Para destino, se evalua principalmente en TRANSFER y CASH_OUT, donde el
    # saldo destino suele tener mayor sentido operativo.
    df["regla_inconsistencia_destino"] = (
        df["type"].isin(["TRANSFER", "CASH_OUT"])
        & (df["diferencia_balance_destino"] > tolerancia)
    ).astype(int)

    df["regla_inconsistencia_balance"] = (
        (df["regla_inconsistencia_origen"] == 1)
        | (df["regla_inconsistencia_destino"] == 1)
    ).astype(int)

    reglas_base = [
        "regla_monto_alto_transfer",
        "regla_ratio_alto",
        "regla_vaciamiento",
        "regla_inconsistencia_balance",
    ]

    # Cantidad de senales base activadas antes de crear la regla compuesta.
    df["cantidad_flags"] = df[reglas_base].sum(axis=1)

    # Regla 5: combinacion de multiples senales.
    # Una transaccion con 3 o mas senales activas merece mayor atencion.
    df["regla_multiples_senales"] = (df["cantidad_flags"] >= 3).astype(int)

    return df


datos_riesgo = aplicar_reglas_riesgo(datos_riesgo, umbrales)

columnas_reglas = [
    "regla_monto_alto_transfer",
    "regla_ratio_alto",
    "regla_vaciamiento",
    "regla_inconsistencia_balance",
    "regla_multiples_senales",
    "cantidad_flags",
]

datos_riesgo[columnas_reglas].mean().sort_values(ascending=False).to_frame("proporcion_activacion")
```

**Interpretacion:** no todas las reglas deben activarse con la misma frecuencia.

Una regla muy frecuente puede indicar una condicion operativa comun, mientras que una regla menos frecuente puede ser mas especifica. La combinacion de senales suele ser mas informativa que una condicion aislada.

## 4. Construccion del score de riesgo

El `risk_score` resume las reglas activadas en un puntaje unico.

Los pesos no estan optimizados estadisticamente. Son una propuesta inicial, interpretable y ajustable, pensada para simular una primera logica de monitoreo basada en negocio.

```python
# Pesos definidos por criterio de negocio.
# Mayor peso implica que la senal se considera mas relevante para riesgo.
PESOS_REGLAS = {
    "regla_monto_alto_transfer": 20,
    "regla_ratio_alto": 20,
    "regla_vaciamiento": 25,
    "regla_inconsistencia_balance": 20,
    "regla_multiples_senales": 15,
}


def calcular_score_riesgo(datos: pd.DataFrame, pesos: dict) -> pd.DataFrame:
    """Calcula el score de riesgo sumando los pesos de reglas activadas."""
    df = datos.copy()
    df["risk_score"] = 0

    for regla, peso in pesos.items():
        df["risk_score"] += df[regla] * peso

    return df


datos_riesgo = calcular_score_riesgo(datos_riesgo, PESOS_REGLAS)
datos_riesgo[["risk_score"] + list(PESOS_REGLAS.keys())].head()
```

**Interpretacion:** el score permite ordenar transacciones segun la cantidad e importancia de senales activadas.

Este puntaje no representa una probabilidad de fraude. Es una medida heuristica de riesgo que ayuda a priorizar revision y analisis.

## 5. Clasificacion de riesgo

El score se transforma en categorias para facilitar la lectura de negocio.

Los umbrales se definen de forma simple:

- Bajo: menos de 20 puntos,
- Medio: entre 20 y 39 puntos,
- Alto: entre 40 y 59 puntos,
- Critico: 60 puntos o mas.

Estas categorias son interpretables y pueden ajustarse posteriormente con evidencia del negocio.

```python
def clasificar_nivel_riesgo(datos: pd.DataFrame) -> pd.DataFrame:
    """Clasifica cada transaccion segun su puntaje de riesgo."""
    df = datos.copy()

    condiciones = [
        df["risk_score"] < 20,
        df["risk_score"].between(20, 39),
        df["risk_score"].between(40, 59),
        df["risk_score"] >= 60,
    ]
    categorias = ["Bajo", "Medio", "Alto", "Critico"]

    df["risk_level"] = np.select(condiciones, categorias, default="Sin clasificar")
    df["risk_level"] = pd.Categorical(
        df["risk_level"],
        categories=categorias,
        ordered=True,
    )

    return df


datos_riesgo = clasificar_nivel_riesgo(datos_riesgo)
datos_riesgo["risk_level"].value_counts().sort_index().to_frame("cantidad")
```

**Interpretacion:** la clasificacion convierte el score en una lectura accionable.

Una transaccion en nivel critico no se declara automaticamente como fraude, pero si deberia tener prioridad en revision o monitoreo.

## 6. Generacion de tabla final

La tabla final consolida variables originales clave, variables derivadas, reglas activadas, score, nivel de riesgo y la marca real `isFraud`.

Esta tabla queda preparada para analisis posterior, dashboard o una futura fase de modelamiento.

```python
# Columnas originales relevantes para trazabilidad y analisis.
columnas_originales = [
    "step",
    "type",
    "amount",
    "nameOrig",
    "oldbalanceOrg",
    "newbalanceOrig",
    "nameDest",
    "oldbalanceDest",
    "newbalanceDest",
]

if "isFlaggedFraud" in datos_riesgo.columns:
    columnas_originales.append("isFlaggedFraud")

columnas_derivadas_finales = [
    "ratio_monto_saldo",
    "flag_vaciamiento",
    "diferencia_balance_origen",
    "diferencia_balance_destino",
    "flag_monto_alto",
    "flag_transfer",
    "cantidad_flags",
]

columnas_reglas_finales = list(PESOS_REGLAS.keys())
columnas_resultado = ["risk_score", "risk_level", "isFraud"]

columnas_tabla_final = (
    columnas_originales
    + columnas_derivadas_finales
    + columnas_reglas_finales
    + columnas_resultado
)

tabla_riesgo = datos_riesgo[columnas_tabla_final].copy()
tabla_riesgo.head()
```

**Interpretacion:** esta tabla funciona como una capa analitica intermedia.

No reemplaza el dataset original, sino que lo enriquece con senales de riesgo y una clasificacion interpretable.

### Variables que explican el fraude

Las visualizaciones anteriores muestran como se distribuye el score y como se clasifican las transacciones por nivel de riesgo.

En este bloque el foco cambia: se comparan directamente las principales variables de riesgo entre transacciones fraudulentas y no fraudulentas. El objetivo es identificar que señales presentan mayor diferencia entre ambos grupos y, por lo tanto, ayudan a explicar mejor el comportamiento del fraude.

```python
# Comparacion de variables de riesgo entre fraude y no fraude.
# Para flags binarios, el promedio equivale a la proporcion de activacion.
variables_comparacion = {
    "Ratio monto/saldo": "ratio_monto_saldo",
    "Cantidad de flags": "cantidad_flags",
    "Vaciamiento de cuenta": "flag_vaciamiento",
    "Monto alto": "flag_monto_alto",
    "Transferencia": "flag_transfer",
    "Inconsistencia balance": "regla_inconsistencia_balance",
}

resumen_variables = []

for nombre_variable, columna in variables_comparacion.items():
    if columna not in tabla_riesgo.columns:
        continue

    resumen = (
        tabla_riesgo.groupby("isFraud", observed=True)[columna]
        .mean()
        .rename(index={0: "No fraude", 1: "Fraude"})
    )

    valor_no_fraude = resumen.get("No fraude", np.nan)
    valor_fraude = resumen.get("Fraude", np.nan)

    resumen_variables.append(
        {
            "variable": nombre_variable,
            "No fraude": valor_no_fraude,
            "Fraude": valor_fraude,
            "diferencia_abs": abs(valor_fraude - valor_no_fraude),
        }
    )

comparacion_variables = pd.DataFrame(resumen_variables)
comparacion_variables = comparacion_variables.sort_values("diferencia_abs", ascending=False)

comparacion_larga = comparacion_variables.melt(
    id_vars=["variable", "diferencia_abs"],
    value_vars=["No fraude", "Fraude"],
    var_name="clase",
    value_name="promedio_o_proporcion",
)

plt.figure(figsize=(10, 5))
ax = sns.barplot(
    data=comparacion_larga,
    x="variable",
    y="promedio_o_proporcion",
    hue="clase",
    palette={"No fraude": "#4C78A8", "Fraude": "#E45756"},
    order=comparacion_variables["variable"],
)
ax.set_title("Comparacion de variables de riesgo: fraude vs no fraude")
ax.set_xlabel("Variable de riesgo")
ax.set_ylabel("Promedio o proporcion")
ax.legend(title="Clase")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()

comparacion_variables
```

**Interpretacion de negocio:** este grafico permite observar que señales diferencian con mayor claridad a las transacciones fraudulentas de las normales.

Las variables con mayor distancia entre ambas barras son las que aportan una lectura mas fuerte del comportamiento fraudulento. Si el fraude muestra mayor cantidad de flags, mayor frecuencia de vaciamiento, mayor presencia de transferencias o mayor proporcion de montos altos, la logica de reglas es coherente con los patrones observados en el EDA.

Desde una perspectiva de negocio, estas diferencias sugieren que el fraude no se explica por una sola condicion aislada. La combinacion de monto, tipo de transaccion, consumo del saldo e inconsistencias operativas entrega una lectura mas completa del riesgo transaccional.

### Activacion de reglas en transacciones fraudulentas

Para complementar la lectura del score, se revisa que reglas se activan con mayor frecuencia dentro de las transacciones marcadas como fraude.

Esta vista ayuda a entender que senales aportan mas informacion en los casos fraudulentos reales.

```python
# Proporcion de activacion de reglas en fraudes reales.
reglas_score = list(PESOS_REGLAS.keys())
fraudes_reales = tabla_riesgo[tabla_riesgo["isFraud"] == 1]

activacion_reglas_fraude = (
    fraudes_reales[reglas_score].mean().sort_values(ascending=False) * 100
)

plt.figure(figsize=(9, 4))
ax = sns.barplot(
    x=activacion_reglas_fraude.values,
    y=activacion_reglas_fraude.index,
    color="#E45756",
)
ax.set_title("Activacion de reglas en transacciones fraudulentas")
ax.set_xlabel("Activacion en fraudes (%)")
ax.set_ylabel("Regla de riesgo")
plt.show()

activacion_reglas_fraude.to_frame("activacion_en_fraudes_%")
```

**Interpretacion de negocio:** esta vista permite identificar que reglas explican mejor los fraudes observados.

Si reglas como vaciamiento, ratio alto o monto alto aparecen con frecuencia en fraudes reales, se refuerza su utilidad para priorizacion. Si alguna regla se activa poco, puede mantenerse como senal complementaria o revisarse en una iteracion posterior.
